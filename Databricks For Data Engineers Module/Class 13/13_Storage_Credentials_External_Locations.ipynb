# Storage Credentials VS External locaitons


**Storage Credentials**

- Defines an authentication mechanism 
- In azure, can use an access connector managed identity 
- can be referenced by external locations 
- Does not itself defines which individual files users can access


**External Location**
- Defines a cloud storage URL 
- Refrences a storage credential
- Can be used to govern external tables, external voluymes and file access
- Supports unity catalog pivileges.
