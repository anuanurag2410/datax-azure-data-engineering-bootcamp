# What is a Dataframe?

- DF is a Spark's Structured representation of data
- Spark uses it while procesing the data
- It is in rows and columns format
- we perform transformations on it.
- It is **not atuotmatically permanent storage**

- Creating a DataFrame does not automatically create a table.
- Creating a DataFrame does not automatically persist your transformed data.

# Introducing PAQRUET

- Parquet is a columnar file format.
- Instead of thinking about the data only row-by-row, it stores data in a way optimized for column-oriented analytical processing.
- It preserves schema information.
- It supports efficient compression.
- It works very well with Spark.


```python
parquet_path='/Volumes/dataxdevelopment/default/datax'


orders_df.write \
    .mode("overwrite") \
    .parquet(parquet_path)
```

```python
parquet_df=spark.read.parquet("/Volumes/dataxdevelopment/default/datax")
parquet_df.show()
```

# INTRODUCTING DELTA 

**What is Delta?**

- Delta Lake is a storage layer commonly used in Databricks
- Delta data uses the Parquet Data plus the Transaction Metadata
- This enables the capabilities that plain **Parquet** doesn't have or provide

_The properties that DELTA Provides_

- ACID
- Transaction Logs
- Time Travel 
- RESTORE 
- MERGE

```python
delta_path="/Volumes/dataxdevelopment/default/delta_datax"
orders_df.write.format("delta").mode('overwrite').save(delta_path)
orders_df.write.format("delta").mode('overwrite').saveAsTable('default.orders_delta')
```
